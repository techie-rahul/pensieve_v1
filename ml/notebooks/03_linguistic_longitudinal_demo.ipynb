```python
import sys
import os

# Ensure project root is on sys.path
project_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from ml.linguistic import extract_linguistic_features
from ml.longitudinal import analyze_journal_history
print("Phase 3 modules imported successfully.")
```

```python
# 1. Linguistic Feature Extraction
sample_entry = "I feel so deeply grateful for my quiet morning routine! Did you remember to stretch? Yes, I did."
features = extract_linguistic_features(sample_entry)

print("Linguistic Features:")
for key, val in features.items():
    if key != "pos_distribution":
        print(f"  {key}: {val}")
print("\nTop POS tags:", {k: v['count'] for k, v in list(features['pos_distribution'].items())[:5]})
```

```python
# 2. Longitudinal History Analysis Across Weekly Windows
from ml.phase3_demo import QUALITATIVE_JOURNAL_HISTORY, THEME_NAMES

results = analyze_journal_history(QUALITATIVE_JOURNAL_HISTORY, window_type="weekly", theme_names=THEME_NAMES)

print(f"Status: {results['status']}")
print(f"Total Entries: {results['entry_count']} spanning {results['time_span_days']} days")
print(f"Windows Analyzed: {results['num_windows']}\n")

print("Emotion Trend Observations:")
for obs in results["emotion_trends"]["observations"]:
    print(f"  -> {obs}")

print("\nTheme Dynamics:")
for obs in results["theme_trends"]["observations"]:
    print(f"  -> {obs}")
```

```python
# 3. Theme-Emotion Associations & Safeguards
print("Theme-Emotion Associations:")
for assoc in results["theme_emotion_associations"]["associations"]:
    print(f"  - [{assoc['theme_name']}]: {assoc['summary']}")

print("\n--- Demonstrating Insufficient Data Safeguard (<3 entries) ---")
sparse = QUALITATIVE_JOURNAL_HISTORY[:2]
sparse_res = analyze_journal_history(sparse)
print(f"Status: {sparse_res['status']}")
print(f"Message: \"{sparse_res['message']}\"")
```
