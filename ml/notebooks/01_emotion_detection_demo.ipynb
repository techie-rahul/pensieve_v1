# Pensieve ML - Phase 1: Emotion Detection Demonstration

> **IMPORTANT NOTICE**: The journal entries analyzed in this notebook are simulated qualitative demonstration examples designed to test realistic journaling scenarios. They are **NOT** benchmark evaluation data.

This notebook demonstrates how the Pensieve emotion detection pipeline classifies multi-label affective states from user reflections using a fine-tuned RoBERTa model (`SamLowe/roberta-base-go_emotions`).

```python
import sys
import os
# Ensure project root is on sys.path
project_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from ml.emotion.inference import EmotionClassifier
from ml.emotion.labels import EMOTION_LABELS, POSITIVE_EMOTIONS, NEGATIVE_EMOTIONS

print(f"Total GoEmotions categories: {len(EMOTION_LABELS)}")
```

```python
# Initialize classifier with validation-tuned threshold
classifier = EmotionClassifier(default_threshold=0.30)
print(f"Initialized EmotionClassifier on device: {classifier.device}")
```

```python
# Qualitative Journal Examples
journal_entries = [
    "Woke up early, drank warm coffee while watching the sunrise. I feel so peaceful and deeply grateful.",
    "I have a huge presentation tomorrow and my chest is tight with dread. What if I forget everything?",
    "Another two-hour meeting with zero actionable outcomes. Completely frustrated with this team process.",
    "We finally launched the v1 platform today! I am so happy and thrilled with the whole team's dedication.",
    "Walked past our old favorite diner and felt an unexpected wave of sorrow missing my best friend."
]

results = classifier.predict_batch(journal_entries)
for r in results:
    print(f"Entry: \"{r['text']}\"")
    print("Detected Emotions:")
    for p in r["predictions"]:
        print(f"  -> {p['emotion']} (confidence: {p['score']:.4f}, sentiment: {p['sentiment']})")
    print("-" * 70)
```
